```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sentence_transformers import SentenceTransformer
from sklearn.neighbors import NearestNeighbors
import pickle
import requests
import time
```

```python
df = pd.read_csv('C:/Users/hp333/Desktop/Recommander_system/model/data/movies_databse.csv')
df.columns
```

```python
df.head()
```

```python
dataforgettingid = df[['number', 'title']].copy()
dataforgettingid
```

```python
vectors = CountVectorizer(stop_words='english', max_features=1000)
vectorized_data = vectors.fit_transform(df['mix'], df['year'])
vectorized_data = vectorized_data.toarray()
```

```python
vectors.get_feature_names_out()
```

```python
knn = NearestNeighbors(n_neighbors = 20, metric = 'cosine')
```

```python
model = knn.fit(vectorized_data)
```

```python
dis, ind = knn.kneighbors(vectorized_data[0].reshape(1, -1))
dataforgettingid.iloc[ind[0], :].head(20)
```

```python
df = df.reset_index(drop=True)
```

```python
with open('model.pkl', 'wb') as f:
    pickle.dump(
        {
            'df' : dataforgettingid,
            'vector' : vectorized_data,
            'model' : model
        }, f
    )
```

```python
transformer = SentenceTransformer('all-MiniLM-L6-v2')
BERT_vector = transformer.encode(df['mix'].astype(str), show_progress_bar = True, device = 'cuda')
```

```python
knn_model = knn.fit(BERT_vector)
```

```python
import pickle

with open('model.pkl', 'wb') as f:
    pickle.dump(
        {
            'df' : df,
            'BERT_vector' : BERT_vector,
            'model' : knn_model
        }, f
    )
```

```python
def get_poster_url(imdb_id, api_key):
    # Step 1: Use the IMDb ID to find the movie in TMDb
    url = f"https://api.themoviedb.org/3/find/{imdb_id}?api_key={api_key}&external_source=imdb_id"
    time.sleep(0.5)
    response = requests.get(url, verify=True)
    data = response.json()

    # Step 2: Check if we got any results
    if data.get("movie_results"):
        poster_path = data["movie_results"][0].get("poster_path")
        
        # Step 3: Build the full poster URL
        if poster_path:
            return f"https://image.tmdb.org/t/p/w500{poster_path}"
    
    return None
```

```python
def recommand(movie_name: str, n: int):
    # api_key = "579df19b2fbced119afad02b8268bfb3"
    recommanded_movies = []
    ids = []
    movies_urls = []
    # Find the index of the movie by title
    if movie_name not in df['title'].values:
        print(f"Movie '{movie_name}' not found in the dataset.")
        return None
    movie_idx = df[df['title'] == movie_name].index[0]
    # Get the nearest neighbors
    dis, ind = knn.kneighbors(BERT_vector[movie_idx].reshape(1, -1), n_neighbors=n+1)
    for i in ind[0]:
        if i != movie_idx:
            recommanded_movies.append(df.iloc[i][1])
            ids.append(df.loc[i][0])
    #poster fetching
    # for i in ids:
    #     poster_url = get_poster_url(i, api_key)
    #     movies_urls.append(poster_url)
    return recommanded_movies, ids
```
